# StainNormProgram Macenko

StainNormProgram ships with Macenko's normalization. This tutorial describes how to customize the MacenkoPipeline with specific parameters.

```python
from stainnormprogram.norm import Normalization
from stainnormprogram.norm.pipeline.macenko import MacenkoPipeline
```

## Using thumbnails instead of tiles

Instead of using tiles of the highest resolution for calculations, you can use thumbnails instead, which may be faster.

```python
N = Normalization()
N.set_target("TCGA-64-1677-01Z-00-DX1.7720D42B-1F57-4C23-9BC5-87C456E7737B.svs")
N.set_source("TCGA-LL-A6FP-01Z-00-DX1.6261398A-7288-4924-BBE2-FC1949256E40.svs")

# Create a MacenkoPipeline and set the style estimation to thumbnail
MP = MacenkoPipeline(
    style_est_mode="thumbnail"
)
N.set_pipeline(MP)

N.run_norm("./output/")
```

## IHC normalization

StainNormProgram supports immunohistochemistry (IHC) normalization with diaminobendizine (DAB) as the chromogen and hematoxylin as the counterstain. By default, MacenkoPipeline is for hematoxylin and eosin (H&E), so you need to change the stain parameter for IHC.

```python
N = Normalization()
N.set_target("your_ihc_target.svs")
N.set_source("TumorCenter_CD3_block1.svs")

MP = MacenkoPipeline(
    stain="DAB"
)
N.set_pipeline(MP)

N.run_norm("./output/")
```

## Change the alpha and beta parameters

You can change the alpha and beta parameters as defined in Macenko's paper.

```python
N = Normalization()
N.set_target("TCGA-64-1677-01Z-00-DX1.7720D42B-1F57-4C23-9BC5-87C456E7737B.svs")
N.set_source("TCGA-LL-A6FP-01Z-00-DX1.6261398A-7288-4924-BBE2-FC1949256E40.svs")

MP = MacenkoPipeline(
    alpha=0.1,
    beta=0.1
)
N.set_pipeline(MP)

N.run_norm("./output/")
```
