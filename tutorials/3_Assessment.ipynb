# StainNormProgram Assessment

After normalization, we can perform an assessment of the images.

```python
from stainnormprogram.assessment import Assessment
```

When you create an Assessment, specify a directory where the assessment metrics and metadata will be saved.

```python
A = Assessment("./assessment/")
```

## Adding a Reference

To perform an Assessment, you need to specify a **reference**. All metrics will be relative to the reference.

**What makes a good reference?**
* **Your target**: if you are interested in how far the normalized images are from the target style
* **Past normalized images**: if you are interested in how much your new normalized images have drifted from your already normalized set
* **All of your normalized images**: if you are interested in how far each image is away from the average of all normalized images

When you call `add_images` set `reference=True` or provide a list of image IDs (filename without extension) to make these images your reference.

```python
A.add_images("target.svs", reference=True)
```

## Adding Images

You can then provide Assessment with the images to assess.

We support grouping the images to keep things organized. An image can be placed in a **group**. For example, you could group images by source (e.g., a group of Hospital A images and a group of Hospital B images).

You can also specify the **version** of images. Often, you will want to assess images before and after normalization. You could then specify an "Original" version and a "Normalized" version.

Note that some metrics (i.e., contrast and structure preservation) require an original (before) and normalized (after) version. To run these metrics, you need to set `original_version=True` in `add_images` for the version that is considered your original version, and you need at least one version that isn't `original_version=True`.

## Compute Assessment

Assessment has several criteria that you can compute:
* **Color distance ("color_dist")**: Bhattacharyya distance between color (HSV hue, HSV saturation, HSV value, joint RGB, joint OKLAB) histograms.
* **Palette ("palette")**: Distances between color palettes computed using modified median cut quantization.
* **Gamut ("gamut")**: Colors outside of the reference gamut.
* **Background ("background")**: Background color distribution.
* **Structure preservation ("structure")**: Preservation of Canny edges before and after normalization.
* **Contrast ("contrast")**: RMS contrast difference before and after normalization.

```python
# To compute all metrics
A.compute()

# To compute only some metrics
A.compute(criteria=["color_dist", "palette"])
```

## Show Plots

After computing the Assessment, you can show the results as interactive Plotly plots. For more details as to the parameters that can be passed in, please refer to the API documentation.

```python
# To show the color distance plot
# plot_type can be "h" for HSV hue, "s" for HSV saturation, "v" for HSV value, "rgb" for RGB, or "oklab" for OKLAB
A.color_distance(plot_type="h", clip_ids=True, sort_by="Norm")

# To show the palette plot
# plot_type can be "compare" for pairwise palette comparisons, "Hdiv" for hematoxylin diversity, "Ediv" for eosin diversity, or "HEsep" for hematoxylin and eosin separation
A.palette(plot_type="compare", clip_ids=True, sort_by="Norm")

# To show the gamut plot
# plot_type can be "oog_count" for out-of-gamut color count or "ig_count" for in-gamut color count
A.gamut(plot_type="oog_count", clip_ids=True, sort_by="Norm")

# To show the background plot
A.background(clip_ids=True, sort_by="Norm")

# To show the contrast plot
A.contrast(clip_ids=True, sort_by="Norm")

# To show the structure preservation plot
A.structure(clip_ids=True, sort_by="Norm")
```

## Export

You can export your results either as Matplotlib figures (Plotly saving is a bit complicated) or as numbers in a CSV.

```python
# To save the color distance plot
A.color_dist(plot_type="h", clip_ids=True, sort_by="Norm", save_path="./color_dist_figure.png")

# To save metrics as a CSV
A.write_to_csv()
```
