# StainNormProgram Basics

This tutorial demonstrates the basic functionality of StainNormProgram. To get the example images used here, see the [library for instructions](https://github.com/peterhsu-uiuc/StainNormProgram/tree/main/src/stainnormprogram/library/library.md).

```python
from stainnormprogram.norm import Normalization
```

```python
N = Normalization()
```

In stain normalization, there are two types of images.

* **Source** images are the ones that will be transformed.
* **Target** images contain the color style that you want the Source images to have.


`set_target` allows you to add target images. Targets can be:

1. A path to an image
2. A path to a directory of images
3. A list of image paths or directories
4. A path to a target style JSON

```python
N.set_target("TCGA-64-1677-01Z-00-DX1.7720D42B-1F57-4C23-9BC5-87C456E7737B.svs")
```

`set_source` allows you to add source images. Sources can be:

1. A path to an image
2. A path to a directory of images
3. A Subset
4. A list of any above

Subsets are for complex workflows. See `1_Inputs.ipynb` for more details.

```python
N.set_source([
    "TCGA-LL-A6FP-01Z-00-DX1.6261398A-7288-4924-BBE2-FC1949256E40.svs",
    "TCGA-LL-A6FR-01Z-00-DX1.107010FD-8EDD-406D-B4DE-DEC9E85576FF.svs",
])
```

`Pipeline`s describe the stain normalization method. StainNormProgram currently supports several variants of Macenko's method which allow you to customize certain settings. For greater customization of the MacenkoPipeline, see `3_MacenkoPipeline.ipynb`.

```python
N.set_pipeline("MacenkoPipeline")
```

We can print a representation of the Normalization object, which shows an overview of what we have specified so far.

```python
print(N)
```

We can now run stain normalization. Specify an `output_dir` path to output files to. This directory should not already exist as this function will create it for you.

```python
N.run_norm("./output/")
```
